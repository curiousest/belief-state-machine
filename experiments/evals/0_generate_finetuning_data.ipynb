# overview
Overall, I want to write science fiction short stories really fast and effectively.
My method is:

1. Come up with idea
2. Write an outline that covers: 
- beliefs that are changed or touched on in the story
- the audience who holds those beliefs
- the story outline
- the story setting
- the characters

3. Use inspired works to initialize writing
a) Use the outline to identify authors/pieces in literature that the work could draw inspiration from.
b) Find a text version of the work online. Select portions of the text for more intentional inspiration (model fine-tuning or few-shot prompts)
c) Generate text from the model inspired by works, as well as specific prompts for setting, exposition, dialogue, scene opening, character descriptions, etc.
d) Filter for "good" text, phrases, paragraphs, etc.

4. Manually assemble the text into one coherent piece.
