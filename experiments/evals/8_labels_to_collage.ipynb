```python
import pandas as pd
import datetime
import os
```

```python
# Change this
PROJECT = "adam_and_eve"

LABEL_FOLDER = "labels/generate_writing/"
TARGET_FOLDER = f"generated_text/{PROJECT}/"
file_names = sorted([f for f in os.listdir(LABEL_FOLDER) if PROJECT in f])
file_names
```

```python
# Change this line
target_files = file_names[:1]

dfs = []
for file in target_files:
    df = pd.read_parquet(f"{LABEL_FOLDER}/{file}")    
    dfs.append(df)
    
combined_df = pd.concat(dfs, ignore_index=True)
target_df = combined_df[(df["target_text"] != df["text"]) & (df["target_text"] != "")]
target_df = target_df.sort_values(by=["label", "usage_text"], ascending=[True, False])

try:
    os.mkdir(f"{TARGET_FOLDER}filtered/")
except FileExistsError: pass
    
def dta(df, val):
    txt = '\n\n---\n\n'.join(df[df['label']==val]['target_text'].astype(str))
    with open(f"{TARGET_FOLDER}filtered/{val}_{datetime.datetime.now()}.txt", "w") as file:
        file.write(txt)

dta(target_df, 'amazing')
dta(target_df, 'ok')
dta(target_df, 'bad')
```
