```python
import os
from pandas import pandas as pd
from tqdm import tqdm

```

```python
PROJECT_NAME = "Assembly_Beyond"
EXPERIMENT_NAME = "Assembly_Beyond"

directory = f"./generated_text/{PROJECT_NAME.replace(' ', '-')}/{EXPERIMENT_NAME.replace(' ', '-')}/"
folders = [name for name in os.listdir(directory) if os.path.isdir(os.path.join(directory, name))]

with open(directory + 'system.txt', 'r') as f:
    system = f.read()
with open(directory + 'user.txt', 'r') as f:
    user = f.read()
with open(directory + 'model.txt', 'r') as f:
    model = f.read()

df = pd.DataFrame()
counter = 0

for folder in tqdm(folders):
    folder_path = directory + folder
    files = [name for name in os.listdir(folder_path) if os.path.isfile(os.path.join(folder_path, name)) and name not in ['system.txt', 'user.txt']]
    for file in files:
        with open(folder_path + file, 'r') as f:
            text = f.read()
        df.append({
            'system_prompt': system,
            'user_prompt': user,
            'model': model,
            'text': text,
            'batch_id': folder,
            'temperature': float(file[:-4]),
            'project_name': PROJECT_NAME,
            'experiment_name': EXPERIMENT_NAME
        })
        counter +=1
df.to_parquet(f'{directory}/output_file.parquet', engine='pyarrow')
print(f'Created with {counter} entries.')
            
        
```
