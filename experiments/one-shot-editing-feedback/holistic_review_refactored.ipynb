# Holistic Review - Refactored

This notebook demonstrates the refactored analysis engine for full manuscript analysis.

```python
# Import the modular components
from analysis_engine import ManuscriptAnalyzer, AnalysisConfig
from prompts import prompt_library
from output_manager import OutputManager
```

```python
# Configuration
config = AnalysisConfig(
    model="gpt-4-0125-preview",
    api_delay=2.0,
    temperature=0.7
)

# Initialize components
analyzer = ManuscriptAnalyzer(config)
output_manager = OutputManager()
```

```python
# Load manuscript
manuscript_path = "/Users/douglashindson/workspace/blog/tabum/outputs/output-2025-03-22.md"
text = analyzer.load_manuscript(manuscript_path)

# Get manuscript info
info = analyzer.get_manuscript_info(text)
print(f"Manuscript Info:")
print(f"  Tokens: {info['token_count']:,}")
print(f"  Chapters: {info['chapter_count']}")
print(f"  Est. Pages: {info['estimated_pages']}")
```

```python
# Get prompts - you can use all holistic prompts or select specific ones
prompts = prompt_library.holistic_analysis_prompts

# Or select specific prompts:
# prompts = prompt_library.get_custom_prompts([
#     "structure_division_meaning.txt",
#     "themes_embodiment.txt",
#     "revision_opportunities.txt"
# ])

print(f"Using {len(prompts)} analysis prompts:")
for name, _ in prompts:
    print(f"  - {name}")
```

```python
# Estimate time
estimated_minutes = len(prompts) * config.api_delay / 60
print(f"\nEstimated analysis time: {estimated_minutes:.1f} minutes")
print("Starting holistic analysis...")
```

```python
# Perform holistic analysis
results = analyzer.analyze_full_manuscript(text, prompts)
print("\nAnalysis complete!")
```

```python
# Save results
output_folder = output_manager.save_holistic_analysis(results, manuscript_path)
output_manager.save_results_json(output_folder, results)

print(f"Results saved to: {output_folder}")

# Show summary
error_count = sum(1 for r in results if r.error)
success_count = len(results) - error_count
print(f"Summary: {success_count} successful, {error_count} errors")
```

```python
# Optional: Display results summary
print("\nAnalysis Results:")
for result in results:
    status = "✅" if not result.error else "❌"
    print(f"  {status} {result.prompt_name}")
    if result.error:
        print(f"      Error: {result.error}")
```

```python
# Optional: Show a sample result
if results and not results[0].error:
    sample = results[0]
    print(f"\nSample result - {sample.prompt_name}:")
    print(f"Bad Cop (first 300 chars): {sample.bad_cop_response[:300]}...")
    print(f"\nGood Cop (first 300 chars): {sample.good_cop_response[:300]}...")
```
